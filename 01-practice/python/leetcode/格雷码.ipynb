```python
def graycode(n,k):
    graycodelist = [0,1]
    for i in range(1,n+1):
        graycodelist2 = graycodelist.copy()
        for index,value in enumerate(graycodelist2):
            if index <= 2**n:
                graycodelist2[index]=bin("0" + str(graycodelist[index]))
                print(graycodelist2[index])
            if index > 2**n:
                graycodelist2.append(bin("1" + str(graycodelist[index-len(graycodelist2)/2])))
        graycodelist = graycodelist2.copy()
        print(graycodelist2)
        
    return graycodelist[k-1]
    
        
if __name__ == "__main__":
    
    n = input("n= ")
    k = input("k= ")
    
    print(graycode(int(n),int(k)))
```

```python
n, k = map(int, input().split())

res = []
for i in range(n - 1, -1, -1):
    if k < (1 << i):  # 在前一半
        res.append('0')
    else:  # 在后一半
        res.append('1')
        k = (1 << i) * 2 - 1 - k  # 映射到逆序位置

print(''.join(res))
```
